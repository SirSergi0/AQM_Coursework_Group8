```python
import pennylane
from shortalyst.shortalyst.shors_algorithm import shors_algorithm
import jax.numpy as jnp
import numpy as np

# from test_shor import TestShor
```

```python
N = 91
a = np.random.randint(2, N, 5)
n_bits = int(jnp.floor(jnp.log2(N))) + 1
n_trials = 10
for i in a:
  print(i)
  p, q, success_prob, _, _ = shors_algorithm(N, i, n_bits, n_trials)

  print(f"p = {p}, q = {q}, success_prob = {success_prob:.4f}")
  if p != 0 and q != 0:
    break
```
